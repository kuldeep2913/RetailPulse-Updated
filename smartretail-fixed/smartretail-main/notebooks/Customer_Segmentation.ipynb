```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
```

```python
rfm = pd.read_csv("rfm_data.csv")
rfm.head()
```

```python
rfm.info()
```

```python
rfm.describe()
```

```python
X = rfm[["Recency", "Frequency","Monetary"]]
X.head()
```

```python
#Normalize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled[:5]
```

```python
inertia = []
for k in range(1,11):
    model = KMeans(n_clusters=k, random_state=42,n_init= 10)
    model.fit(X_scaled)
    inertia.append(model.inertia_)
```

```python
plt.figure(figsize=(8,5))
plt.plot(range(1,11),inertia,marker ="o")
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("inertia")
plt.show
```

```python
#Kmeans model
kmeans = KMeans(n_clusters=4, random_state=42,n_init =10)
rfm["Cluster"] = kmeans.fit_predict(X_scaled)
rfm.head()
```

```python
#cluster distribution
plt.figure(figsize=(6,5))
sns.countplot(data = rfm,x= "Cluster")
plt.title("Customers in Each Cluster")
plt.show()
```

```python
cluster_summary = (rfm.groupby("CustomerSegment")[["Recency","Frequency","Monetary"]].mean())
cluster_summary
```

```python
cluster_names = {
    0: "Regular",
    1: "At-Risk",
    2: "VIP",
    3: "Loyal"
}

rfm["CustomerSegment"] = rfm["Cluster"].map(cluster_names)
rfm.head()
```

```python
rfm["CustomerSegment"].value_counts()
```

```python
rfm.to_csv("customer_segments.csv", index=False)
print("Customer Segmentation Completed Successfully!")
```
